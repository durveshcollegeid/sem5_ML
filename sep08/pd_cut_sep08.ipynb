Defining manual bins for pd.cut()

```python
df_new["new_age"] = pd.cut( df_new["age"], bins=[0,11,18,45,60], labels=["Young", "Minor", "Adult", "Old"] )
```

```python
df_new.head()
```

Allowing automatic bins for pd.cut()

```python
df_new["auto_age"] = pd.cut( df_new["age"], bins=4, labels=["Young", "Minor", "Adult", "Old"] )
```

```python
df_new.head()
```

Count comparison

```python
df_new["auto_age"].nunique()
```

```python
labels=["Young", "Minor", "Adult", "Old"]

print( "automatic cut" )
for x in labels :
  print( "count of " +x + " = " + str(int(df_new[ df_new["auto_age"] == x ]["auto_age"].count())) )
```

```python
print( "manual cut" )
for x in labels :
  print( "count of " +x + " = " + str(int(df_new[ df_new["new_age"] == x ]["new_age"].count())) )
```
