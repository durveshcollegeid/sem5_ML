# Aggregate function on grouped data

```python
df_grouped_original = df.groupby('Neighborhood')
```

```python
df_grouped_original['SalePrice'].mean()
```

```python
df_grouped_original['SalePrice'].max()
```

# Grouping on multiple feature columns

```python
df_multi_grouped = df.groupby(['Neighborhood', 'Street'] )
```

```python
df_multi_grouped['SalePrice'].mean()
```

```python
df_multi_grouped['SalePrice'].max()
```

```python
df_multi_grouped['SalePrice'].describe()
```

# Aggregation dictionary on multiple data columns using .agg()

```python
df_multidata = df.groupby( "Neighborhood" ).agg(
    {
        "SalePrice": "mean",
        'GarageCars': "max"
    }
)
```

```python
df_multidata
```

```python
df_multidata.describe()
```
