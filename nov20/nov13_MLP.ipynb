# Experiment 1

```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
xor = np.array( [ [0,0],[0,1],[1,0],[1,1] ] )
```

```python
y = np.array( [0,1,1,0] )
```

```python
model_p = Perceptron( max_iter=1000, random_state=42 )
model_p.fit( xor, y )
```

```python
print( "Perceptron predictions : ", model_p.predict( xor ) )
print( "Perceptron accuracy : ", accuracy_score( y, model_p.predict( xor ) ) )
```

```python
model_mlp = MLPClassifier( hidden_layer_sizes=(4,),
activation = "tanh", solver="lbfgs", max_iter=5000, random_state=42
                           )
```

```python
model_mlp.fit( xor, y )
```

```python
print( "MLP predictions : ", model_mlp.predict( xor ) )
print( "MLP accuracy : ", accuracy_score( y, model_mlp.predict( xor ) ) )
```

# Experiment 2

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
```

```python
X, y = make_moons( n_samples=500, noise=0.2, random_state=42 )
```

```python
plt.scatter( X[:,0], X[:,1], c=y )
plt.xlabel( "Feature 1" )
plt.ylabel( "Feature 2" )
plt.title( "Nonlinear Classification - Make Moons" )
plt.show()
```

```python
from sklearn.linear_model import Perceptron
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score
```

```python
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.25, random_state=42, stratify=y )
```

```python
perceptron = make_pipeline(StandardScaler(),Perceptron( max_iter=1000, random_state=42 ))
```

```python
perceptron.fit( X_train, y_train )
```

```python
pred = perceptron.predict( X_test )
```

```python
print( "Perceptron accuracy : ", accuracy_score( y_test, pred ) )
```
